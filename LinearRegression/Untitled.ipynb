```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Book1.csv", sep='\t')
data
```

```python
# data
plt.scatter(data.videos, data.views)
plt.xlabel('Videos')
plt.ylabel('Views')
plt.title('Videos vs Views')
plt.show()
```

```python
x = np.array(data.videos.values)
x
```

```python
y = np.array(data.views.values)
y
```

```python
model = LinearRegression()
model.fit(x.reshape(-1,1),y)
```

```python
new_x = np.array([1000]).reshape(-1,1)
new_x
```

```python
model.predict(new_x)
```

```python
m,c = np.polyfit(x,y,1)
```

```python
plt.scatter(data.videos, data.views)
# plt.xlim(1500)
plt.plot(x,m*x+c, color = "red")

# pediction
plt.axvline(x = 1000 , linestyle=':')
```
